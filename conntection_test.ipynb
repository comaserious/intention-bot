```python
import os
from dotenv import load_dotenv

load_dotenv()  # 같은 폴더의 .env (BASE_URL, API_KEY, MODEL) 읽기

base_url = os.getenv("BASE_URL")
api_key = os.getenv("API_KEY")
model = os.getenv("MODEL")

# 주소·키는 출력하지 않는다 (노트북 출력이 git 에 남지 않도록)
print("BASE_URL set:", bool(base_url))
print("API_KEY set:", bool(api_key))
print("MODEL:", model)
```

```python
from openai import OpenAI

client = OpenAI(
    base_url = base_url,
    api_key = api_key
)
```

```python
response = client.chat.completions.create(
    model=model,
    messages=[
        {"role": "user", "content": "heelo?"}
    ]
)
```

```python
print(response)
```

```python
response = client.responses.create(
    model=model,
    input=[
        {"role": "user", "content": "Hello!"}
    ],
    reasoning={"effort": "low"}
)
print(response)
```
